```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
sns.set_theme(style="whitegrid")
```

**data collection**

```python
df=pd.read_csv('/content/solar_power_output.csv')
df
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe().round(2)
```

```python
fig, axes=plt.subplots(1,5,figsize=(22,4))
for ax, col in zip(axes,df.columns):
  sns.histplot(df[col],kde=True,ax=ax,color="pink")
  ax.set_title(col)
plt.show()
```

**Bivariate analysis**

```python
features = ["temperature", "humidity", "solar_irradiance", "wind_speed"]
fig, axes = plt.subplots(1, 4, figsize=(22, 5))
for ax, col in zip(axes, features):
    sns.regplot(x=df[col], y=df["solar_power_output"], ax=ax,
                scatter_kws={"alpha": 0.4}, line_kws={"color": "red"})
    r = df[col].corr(df["solar_power_output"])
    ax.set_title(f"{col} vs output (r = {r:.3f})")
plt.show()
```

**multivariate analysis**

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(),annot=True, fmt=" .2f", cmap="coolwarm")
plt.title("correlation heatmap")
plt.show()
```

**Data Preprocessing**

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
x=df[["solar_irradiance"]]
y=df["solar_power_output"]
```

**data splitting**

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print(x_train.shape, x_test.shape)
```

model training

```python
model= LinearRegression()
print(model)
model.fit(x_train, y_train)
print("Slope:(m=beta1)",model.coef_[0])
print("Intercept:(c=beta0)", model.intercept_)
```

**model evaluation**

```python
y_train_pred = model.predict(x_train)
print("R2  :",r2_score(y_train, y_train_pred))
print("MAE :",mean_absolute_error(y_train, y_train_pred))
print("MSE :",mean_squared_error(y_train,y_train_pred))
```

final model evaluation

```python
y_pred = model.predict(x_test)
print("R2  :",r2_score(y_test, y_pred))
print("MAE :",mean_absolute_error(y_test, y_pred))
print("MSE :",mean_squared_error(y_test,y_pred))
```

```python
plt.scatter(x_test, y_test, label="Actual")
plt.plot(x_test, y_pred, color="red", label="predicted")
plt.xlabel("solar irradiance")
plt.ylabel("solar_power_output")
plt.legend()
plt.show()
```

save model for deployment

```python
joblib.dump(model, "model.pkl")
print("model saved")
```
